```python
import torch
from torch.nn.functional import cosine_similarity

embeddings = torch.load('passage_embeddings.pt')
norms = torch.norm(embeddings, dim=1)
print(f"Embedding norms - Min: {norms.min():.4f}, Max: {norms.max():.4f}")
is_normalized = torch.allclose(norms, torch.ones_like(norms), atol=1e-3)
print(f"Are embeddings normalized? {is_normalized}")

# Compute similarity matrix
similarities = cosine_similarity(embeddings.unsqueeze(1), embeddings.unsqueeze(0), dim=2)

# Find most similar pairs (excluding self-similarity)
similarities_no_diag = similarities.fill_diagonal_(0)
max_sim_idx = torch.unravel_index(similarities_no_diag.argmax(), similarities_no_diag.shape)
print(f"Most similar passages: {max_sim_idx[0].item()} and {max_sim_idx[1].item()}")
print(f"Similarity score: {similarities[max_sim_idx].item():.4f}")

# Find passages most similar to a specific one (e.g., passage 0)
passage_0_sims = similarities[0]
top_similar = torch.argsort(passage_0_sims, descending=True)[1:6]  # Exclude self
print(f"Top 5 passages similar to passage 0: {top_similar.tolist()}")
```
